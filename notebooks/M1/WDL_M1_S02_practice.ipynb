## Q1 · Rooftop
A 120 m² roof receives **65 mm** of rain. How many **litres** fall on it?

```python
#@title ▶ Check Q1
my_answer = 0 #@param {type:"number"}
correct = 65 * 120
if my_answer == 0:
    no("Type your answer in the box first.")
elif close(my_answer, correct):
    ok(f"Correct: {correct:,.2f}")
else:
    no("Not quite. Hint: Depth (mm) × area (m²) = litres.")
```

## Q2 · Campus
A 30 ha campus receives **40 mm** of rain. How many **m³** fall on it?

```python
#@title ▶ Check Q2
my_answer = 0 #@param {type:"number"}
correct = 40 * 30 * 1e4 / 1e3
if my_answer == 0:
    no("Type your answer in the box first.")
elif close(my_answer, correct):
    ok(f"Correct: {correct:,.2f}")
else:
    no("Not quite. Hint: Convert ha to m² first (1 ha = 10,000 m²), get litres, then divide by 1,000 for m³.")
```

## Q3 · Reservoir

## Q4 · River
A river flows at **250 m³/s** for **one day**. How many **MCM** pass?

```python
#@title ▶ Check Q4
my_answer = 0 #@param {type:"number"}
correct = 250 * 86400 / 1e6
if my_answer == 0:
    no("Type your answer in the box first.")
elif close(my_answer, correct):
    ok(f"Correct: {correct:,.2f}")
else:
    no("Not quite. Hint: Flow × seconds in a day (86,400) gives m³. Divide by 1,000,000 for MCM.")
```
